```python
# Optimizar una búsqueda de cadenas en otra cadena

texto = """ El día comienza gris, como mi estado de ánimo. Me levanto con pesadez y me miro en el espejo: ojeras, ojos sin brillo. Un nudo en la garganta me ahoga las emociones.
El desayuno, sin sabor. Mi mente divaga en pensamientos negativos: culpa, duda, miedo. Un baile macabro que me roba la paz.
El trabajo, un refugio temporal. Me concentro en las tareas para huir de la tormenta emocional. Las horas pasan lentas, luchando contra la marea de sentimientos.
Al final del día, estoy agotado. Vuelvo a casa, buscando paz en la soledad. Pero el silencio me enfrenta al vacío interior. Un grito ensordecedor en la quietud.
En la oscuridad, una luz: un recuerdo feliz, una sonrisa, una palabra de aliento. Me aferro a ese destello como náufrago a un salvavidas. Respiro hondo, sintiendo el aire en mis pulmones. La calma me invade.
Llega la noche y el sueño me abraza. En la madrugada, una esperanza: un nuevo día, una nueva oportunidad. El camino será difícil, pero tengo la fuerza para superar las tormentas. Mañana me levantaré buscando la luz en la oscuridad, la paz en el torbellino.
Es un viaje de autodescubrimiento, un camino hacia la resiliencia. Es aprender a vivir con la vida, aceptando sus vaivenes sin perder mi esencia. Es encontrar belleza en el caos, luz en la oscuridad. Es vivir con la intensidad de las emociones.""" 


positive_words = ["una luz", "feliz", "sonrisa", "aliento", "calma", "superar", "luz", "paz", "resiliencia", "esencia", ] 
negative_words = ["gris", "pesadez", "ojeras", "nudo", "ahoga", "sin sabor", "negativos", "culpa", "duda", "miedo", "macabro", "roba", "tormenta", "difícil", "tormentas", "oscuridad"] 
neutral_words = ["refugio", "aferro", "maravilloso", "vuelvo", "buscando", "destello", "respiro hondo", "sintiendo el aire", "invade", "abraza", "nuevo día", "oportunidad", "fuerza", "vivir", ] 


score = 0 
for positive_word in positive_words:    
    score = score + texto.count(positive_word)    

for negative_word in negative_words:    
    score = score - texto.count(negative_word)
    
for neutral_word in neutral_words:    
    score == 0

if score == 0:    
    result = "Neutro" 
if score > 0:    
    result = "Positivo" 
if score < 0:    
    result = "Negativo"    

print("Analizador de sentimientos\n") 

print("texto: {}\nscore: {} \nresultado: {}".format(texto, score, result))

linea1, linea2, linea3, linea4, linea5, linea6, linea7 = texto.split("\n")
print("Líneas escritas: \n")
print(linea1)
print(linea2)
print(linea3)
print(linea4)
print(linea5)
print(linea6)
print(linea7)

print("La cantidad de caracteres en tu estado es: ", len(texto))
```
