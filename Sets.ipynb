# Métodos para manipulación de sets

### add
Agregar un elemento a un set

```python
my_set = {1, 5, 11, 23}

print(type(my_set))
print(len(my_set))
print(my_set)
```

```python
my_set.add(1)
my_set.add(2)
my_set.add(3)

print(type(my_set))
print(len(my_set))
print(my_set)
```

### clear
Remueve todos los elemntos

```python
my_set = {1, 5, 11, 23}

print(type(my_set))
print(len(my_set))
print(my_set, "\n")

my_set.clear()
print(type(my_set))
print(len(my_set))
print(my_set)
```

### copy
Copia los elementos en un nuevo set

```python
my_set = {1, 5, 11, 23}
my_set_2 = my_set.copy()

print(type(my_set))
print(len(my_set))
print(my_set, "\n")


print(type(my_set_2))
print(len(my_set_2))
print(my_set_2, "\n")

my_set_2.clear()

print(type(my_set))
print(len(my_set))
print(my_set, "\n")


print(type(my_set_2))
print(len(my_set_2))
print(my_set_2, "\n")
```

```python
my_set = {1, 5, 11, 23}
my_set_2 = my_set

print(type(my_set))
print(len(my_set))
print(my_set, "\n")


print(type(my_set_2))
print(len(my_set_2))
print(my_set_2, "\n")

my_set_2.clear()

print(type(my_set))
print(len(my_set))
print(my_set, "\n")


print(type(my_set_2))
print(len(my_set_2))
print(my_set_2, "\n")
```

### difference
Obtiene los valores presentes en el primer set y faltantes en el segundo set

```python
set_a = {1, 3, 5, 7, 9}
set_b = {1, 2, 4, 6, 7, 8}

# Valores que hay en set_b faltantes en set_a
print(set_b.difference(set_a))

# Valores que hay en set_a faltantes en set_b
print(set_a.difference(set_b))
```

### difference_update
Conserva solo los valores diferentes entre el set solicitado y el que es mandado como parámetro

```python
set_a = {1, 3, 5, 7, 9}
set_b = {1, 2, 4, 6, 7, 8}

# Valores que hay en set_b faltantes en set_a
set_b.difference_update(set_a)
print(set_b)

# Valores que hay en set_a faltantes en set_b
set_a.difference_update(set_b)
print(set_a)
```

```python
set_a = {1, 3, 5, 7, 9}
set_b = {1, 2, 4, 6, 7, 8}

# Valores que hay en set_a faltantes en set_b
set_a.difference_update(set_b)
print(set_a)

# Valores que hay en set_b faltantes en set_a
set_b.difference_update(set_a)
print(set_b)
```

## discard
Elimina un elemento del set si está presente

```python
my_set = {"Python", "Javascript", "Go", "Scala"}
print(type(my_set))
print(len(my_set))
print(my_set, "\n")

my_set.discard("Go")
print(type(my_set))
print(len(my_set))
print(my_set, "\n")

# No importa que el valor no esté presente
my_set.discard("test")
print(type(my_set))
print(len(my_set))
print(my_set, "\n")
```

### intersection
Obtiene los valores en común en 2 sets

```python
set_a = {1, 3, 5, 7, 9}
set_b = {1, 2, 4, 6, 7, 8}

print(set_a.intersection(set_b))
print(set_b.intersection(set_a))
```

### intersection_update
Obtiene los valores en omún en dos sets y los asigna al set que manda llamar el método

```python
set_a = {1, 3, 5, 7, 9}
set_b = {1, 2, 4, 6, 7, 8}

print(set_a.intersection_update(set_b))
print(type(set_a))
print(len(set_a))
print(set_a, "\n")

print(type(set_b))
print(len(set_b))
print(set_b, "\n")
```

### isdisjoint
True: si no hay valores compartidos por ambos sets

False: si hay por lo meno sun valor compartido por ambos sets

```python
set_a = {1, 3, 5, 7, 9}
set_b = {1, 2, 4, 6, 7, 8}

print(set_a.isdisjoint(set_b))
```

```python
set_a = {1, 3,}
set_b = {1, 2, 4, 6, 7, 8}

print(set_a.isdisjoint(set_b))
```

```python
set_a = {0, 3,}
set_b = {1, 2, 4, 6, 7, 8}

print(set_a.isdisjoint(set_b))
```

### issubset
True: si todos los valores del set llamado están en el set pasado como parámetro

```python
set_a = {1, 8}
set_b = {1, 2, 4, 6, 7, 8}

print(set_a.issubset(set_b))
```

```python
set_a = {1, 3, 8}
set_b = {1, 2, 4, 6, 7, 8}

print(set_a.issubset(set_b))
```

### issuperset
En cierta forma es opuesto a issubset

Analiza si todos los valores del set pasado como parámetro están en el set llamado

```python
set_a = {1, 8}
set_b = {1, 2, 4, 6, 7, 8}

print(set_b.issuperset(set_a))
```

```python
set_a = {1, 3, 8}
set_b = {1, 2, 4, 6, 7, 8}

print(set_b.issubset(set_a))
```

### pop
Elimina el primer elemento del set

```python
my_set = {1, 2, 4, 6, 7, 8}

print(type(my_set))
print(len(my_set))
print(my_set, "\n")

print(my_set.pop())
print(my_set.pop())
print(my_set.pop(), "\n")

print(type(my_set))
print(len(my_set))
print(my_set)
```

### remove
Elimina el elemento pasado como parámetro; debe estar presente en el set o mandará error

```python
my_set = {1, 2, 4, 6, 7, 8}

print(type(my_set))
print(len(my_set))
print(my_set, "\n")

my_set.remove(4)

print(type(my_set))
print(len(my_set))
print(my_set, "\n")

my_set.remove(3)

print(type(my_set))
print(len(my_set))
print(my_set, "\n")
```

### union
Combina 2 sets

```python
set_a = {1, 3, 8}
set_b = {1, 2, 4, 6, 7, 8}

set_c = set_a.union(set_b)
print(set_c)
```

### update 
Añade los valores del set pasado como parámetro al set llamado

```python
set_a = {1, 3, 8}
set_b = {1, 2, 4, 6, 7, 8}

set_a.update(set_b)
print(set_a)
print(set_b)
```

### symmetric_difference
Obtiene las diferencias en ambos sets

```python
set_a = {1, 3, 8}
set_b = {1, 2, 4, 6, 7, 8}

print(set_a.symmetric_difference(set_b))
print(set_b.symmetric_difference(set_a))
```

### symmetric_difference_update
Obtiene las diferencias en ambos sets y asigna los valores al set llamado

```python
set_a = {1, 3, 8}
set_b = {1, 2, 4, 6, 7, 8}

print(set_a.symmetric_difference_update(set_b))
print(set_a)
print(set_b)
```

## Conversión de Sets

### Cadenas a Set

```python
my_string = "Backend Python"
my_set = set(my_string)

print(type(my_string))
print(my_string, "\n")

print(type(my_set))
print(my_set, "\n")
```

### Lista a Set

```python
my_list = [1,1,1,2,2,3,3,4,4,5]
my_set = set(my_list)

print(type(my_list))
print(my_list, "\n")

print(type(my_set))
print(my_set, "\n")
```

### Tupla a Set

```python
my_tuple = (1,1,1,2,2,3,3,4,4,5)
my_set = set(my_tuple)

print(type(my_tuple))
print(my_tuple, "\n")

print(type(my_set))
print(my_set, "\n")
```

### Set a Cadena

```python
my_set = {'P', 'e', 'a', ' ', 'k', 'n', 'y', 'c', 'd', 'o', 't', 'B', 'h'} 
my_string = "".join(my_set)

print(type(my_set))
print(my_set, "\n")

print(type(my_string))
print(my_string, "\n")
```

### Set a Lista

```python
my_set = {1, 2, 3, 4, 5} 
my_list = list(my_set)

print(type(my_set))
print(my_set, "\n")

print(type(my_list))
print(my_list, "\n")
```

### Set a Tupla

```python
my_set = {1, 2, 3, 4, 5} 
my_tuple = tuple(my_set)

print(type(my_set))
print(my_set, "\n")

print(type(my_tuple))
print(my_tuple, "\n")
```
